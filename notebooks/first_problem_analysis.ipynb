```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))
```

```python
from src.plotting import set_plot_style, plot_trend, plot_distribution, plot_yoy_changes, compute_summary_stats

set_plot_style()
CLEAN_DIR = ROOT / "data" / "clean"
TABLES_DIR = ROOT / "output" / "tables"
FIGURES_DIR = ROOT / "output" / "figures"
```

# youth_unemployment

```python
df = pd.read_csv('../data/clean/youth_unemp_2009_2025.csv')
```

```python
stats_unemp = compute_summary_stats(
    df=df,
    change_col="change_pp",
    save_path=TABLES_DIR / "youth_unemp_stats.csv",
)
stats_unemp
```

```python
fig, ax = plot_trend(
    df=df,
    title="Stopa bezrobocia młodzieży (15–24 lata) w latach 2009–2025",
    ylabel="Stopa bezrobocia (% aktywnych zawodowo)",
    ylim=(0, 31),
    highlight_period=(2020, 2022),
    annotate_last=True,
    save_path=FIGURES_DIR / "youth_unemp_trend",
)
plt.show()
```

```python
fig, ax = plot_distribution(
    df=df,
    title="Rozkład rocznych stóp bezrobocia młodzieży (15–24 lata), 2009–2025",
    ylabel="Stopa bezrobocia (% aktywnych zawodowo)",
    ylim=(0, 31),
    legend_loc="lower right",
    save_path=FIGURES_DIR / "youth_unemp_distribution",
)
plt.show()
```

```python
fig, axes = plot_yoy_changes(
    df=df,
    title=(
        "Roczne zmiany stopy bezrobocia młodzieży (2010–2025)\n"
        "Wartości > 0: wzrost bezrobocia  |  Wartości < 0: spadek bezrobocia"
    ),
    ylabel="Zmiana rok do roku (pkt proc.)",
    highlight_period=(2019.5, 2022.5),
    label_threshold=0.0,
    save_path=FIGURES_DIR / "youth_unemp_yoy",
)
plt.show()
```
